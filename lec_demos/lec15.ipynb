## Augmenting Arrays

The `for` statement above does simulate the results of five bets and prints the results.  However, we cannot use this information for further computation. So, we want to save what the results are from the `for` loop into an array.
We can use a`for` statement to create an array of results and augment the array each time we go through the loop.

The `append` method in `NumPy` helps us do this. The call `np.append(array_name, value)` evaluates to a new array that is `array_name` augmented by `value`. When you use `append`, keep in mind that all the entries of an array must have the same type.

But often while using `for` loops it will be convenient to change the array when augmenting it. We do this by assigning the augmented array to the same name as the original - essentially adding on our values from the loop each time.
